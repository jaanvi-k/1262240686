```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report

# 1. Load your custom dataset
# To get this path in Colab: click the folder icon on the left,
# right-click your uploaded file, and select "Copy path"
data = pd.read_csv("/content/winequality-red.csv", sep=';')

# 2. Define the column you are trying to predict
# Replace this with the exact column name from your dataset
target_column = "quality"

# 3. Separate the Features (X) from the Target (y)
X = data.drop(target_column, axis=1)
y = data[target_column]

# --- DATA PREPARATION STEP ---
# Gaussian Naive Bayes only understands numbers. If your dataset has
# categorical text (like "Yes"/"No"), we need to convert them to numbers.
# We will also fill in any blank data cells so the model doesn't crash.
X = pd.get_dummies(X)
X = X.fillna(0)
# -----------------------------

# 4. Partition the data: 80% for training, 20% for testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=40
)

# 5. Initialize the Gaussian Naïve Bayes model
model = GaussianNB()

# 6. Train the model on the training partition
model.fit(X_train, y_train)

# 7. Predict the outcomes for the test partition
y_pred = model.predict(X_test)

# 8. Evaluate and print the accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy * 100:.2f}%\n")

print("Detailed Classification Report:")
print(classification_report(y_test, y_pred))
```
