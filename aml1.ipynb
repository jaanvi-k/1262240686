```python
from sklearn.datasets import load_iris, make_blobs
from sklearn.tree import DecisionTreeClassifier
from sklearn.feature_selection import SequentialFeatureSelector
from sklearn.cluster import KMeans
import pandas as pd
import numpy as np

print("Supervised ML & Feature Selection")

data = load_iris()
X = data.data
y = data.target
feature_names = np.array(data.feature_names)

model = DecisionTreeClassifier(random_state=42)

forward_selector = SequentialFeatureSelector(model, n_features_to_select=2, direction='forward')
forward_selector.fit(X, y)
forward_mask = forward_selector.get_support()
forward_selected_names = feature_names[forward_mask]

backward_selector = SequentialFeatureSelector(model, n_features_to_select=2, direction='backward')
backward_selector.fit(X, y)
backward_mask = backward_selector.get_support()
backward_selected_names = feature_names[backward_mask]

results_table = pd.DataFrame({
    'Feature Name': feature_names,
    'Forward Selection (Kept?)': forward_mask,
    'Backward Selection (Kept?)': backward_mask
})

print(results_table.to_string(index=False))

X_important = forward_selector.transform(X)
model.fit(X_important, y)
print(f"\nModel trained successfully using only: {forward_selected_names}!\n")

print("\n Unsupervised ML (Clustering)")

X_unlabelled, _ = make_blobs(n_samples=100, centers=3, random_state=42)

kmeans_model = KMeans(n_clusters=3, random_state=42, n_init='auto')
clusters = kmeans_model.fit_predict(X_unlabelled)

print("Algorithm successfully grouped the data into 3 clusters!")
print("Cluster assignments for all data points:")
print(clusters)
```
