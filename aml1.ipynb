```python
import pandas as pd
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from mlxtend.feature_selection import SequentialFeatureSelector as SFS

# Load Dataset
wine = load_wine()

X = pd.DataFrame(wine.data, columns=wine.feature_names)
y = wine.target

print("Original Features")
print(X.columns)

# Split Dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=42
)

# Classifier
knn = KNeighborsClassifier(n_neighbors=5)

##################################################
# Sequential Forward Selection
##################################################

forward = SFS(
    knn,
    k_features=5,
    forward=True,
    floating=False,
    scoring='accuracy',
    cv=5
)

forward.fit(X_train, y_train)

print("\nSelected Features using Forward Selection")
print(forward.k_feature_names_)

X_train_forward = X_train[list(forward.k_feature_names_)]
X_test_forward = X_test[list(forward.k_feature_names_)]

knn.fit(X_train_forward, y_train)
prediction = knn.predict(X_test_forward)

print("Forward Selection Accuracy:",
      accuracy_score(y_test, prediction))

##################################################
# Sequential Backward Selection
##################################################

backward = SFS(
    knn,
    k_features=5,
    forward=False,
    floating=False,
    scoring='accuracy',
    cv=5
)

backward.fit(X_train, y_train)

print("\nSelected Features using Backward Selection")
print(backward.k_feature_names_)

X_train_backward = X_train[list(backward.k_feature_names_)]
X_test_backward = X_test[list(backward.k_feature_names_)]

knn.fit(X_train_backward, y_train)
prediction = knn.predict(X_test_backward)

print("Backward Selection Accuracy:",
      accuracy_score(y_test, prediction))
```
